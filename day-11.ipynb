# Set and Tuple

```python
# sets are mutable
# sets are declared or are denoted with{}
# sets don't allow redundancy
# sets are unordered
# unique values
```

```python
#tuples are denoted by ()
# tuples are immutable
# tuples allow redundancy
# tuples are ordered
# used to have some constraints or solid variables
```

```python
s = {1,2,3,3,4,5}
print(s)
```

```python
s.add(6)
print(s)
```

```python
s.remove(3)
print(s)
popped_element = s.pop()
print(s)
print(popped_element)
```

```python
s.discard(5)
print(s)
```

```python
for i in s:
    print(i,end=' ')
```

```python
print(2 in s)
```

```python
print(3 in s)
```

```python
# special functions/operation
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}
print('Union :',s1.union(s2))
print('Union :',s1|s2)
```

```python
print('Intersection :',s1.intersection(s2))
print('Intersection :',s1&s2)
```

```python
print('Difference :',s1.difference(s2))
print('Difference :',s1 - s2)
print(s2-s1)
```

```python
print('Symmetric Difference :',s1.symmetric_difference(s2))
print('Symmetric Difference :',s1 ^ s2)
```
