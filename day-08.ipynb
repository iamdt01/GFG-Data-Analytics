# List Comprehension

```python
l = [i for i in range(1,11)]
print(l)
```

```python
s = [i**2 for i in range(1,11)]
print(s)
```

```python
even = [i for i in range(1,11) if i%2 == 0]
odd = [i for i in range(1,11) if i%2!=0]
print(even)
print(odd)
```

```python
# List Comprehension in multidimensional list
```

```python
s = [(i,j) for i in range(1,4) for j in range(1,4)]
print(s)
```

```python
s1 = [[j for j in range(1,4)]for i in range(1,4)]
print(s1)
```

```python
s2 = [j for i in s1 for j in i]
print(s2)
```

# Dictionary

```python
d = {1:'a',2:'b'}
print(d)
```

```python
# dictionary stores data in the form of key value pair
```

```python
d[3] = 'c'
print(d)
```

```python
# all the items
print(d.items())
```

```python
# all the keys
print(d.keys())
```

```python
# all the values
print(d.values())
```

```python
# the key must be distinct
d[3] = 'd'
```

```python
print(d)
```

```python
print(d[1])
```

```python
# we can access the elements or the values from the keys but not vice versa
```

```python
print(d.get(1)) # get() method
```

```python
del d[1]
print(d)
```

```python
d[1] = 'a'
print(d)
```

```python
l = [d[i] for i in d.keys()]
print(l)
```

```python
for i in d.keys():
    print(i,d[i])
```

```python
for k,v in d.items():
    print(k,'--->',v)
```

```python
# check if key is present
print(1 in d)
print(10 in d)
```

```python
print('a' in d) # values are not entertained
```

```python
# updating dictionaries
d_1 = {1:'A',2:'B'}
d_2 = {1:'a',2:'b'}
d_1.update(d_2)
print(d_1)
print(d_2)
```

```python
# merging disctionaries
d_1 = {1:'A',2:'B'}
d_2 = {3:'a',4:'b'}
d_3 = d_1 | d_2
print(d_1)
print(d_2)
print(d_3)
```
