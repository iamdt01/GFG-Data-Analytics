# Control Statements If and else

```python
n = -5
if n < 0:
    print("Negative number")
else:
    print("Non-negative number")
```

```python
n = int(input("Enter a number: "))
if n < 0:
    print("Negative number")
elif n == 0:
    print("Zero")
else:
    print("Non-negative number")
```

```python
# we can write the conditions in the below manner as well
# 1. multiple or single if statements
# 2. single if and a single else statement
# 3. single if, multiple elif and a single else statement
```

```python
n = int(input("Enter a number: "))
if n%5 == 0:
    print(n,"is divisible by 5")
else:
    print(n,"is not divisible by 5")
```

```python
n = int(input("Enter a number: "))
if n%5 == 0 and n%3 == 0:
    print(n,"is divisible by 5 and 3")
elif n%5 == 0:
    print(n,"is divisible by 5 and not 3")
elif n%3 == 0:
    print(n,"is divisible by 3 and not 5")
else:
    print(n,"is not divisible by 5 and 3")
```

# For Loop

```python
n = int(input("Enter a number: "))
for i in range(11):
    print(n,'*',i,'=',n*i)
```

```python
for i in range(1,11,2):
    print(i)
```

```python
for i in range(4,(4*10)+1,4):
    print(i)
```

```python
n = int(input("Enter a number: "))
for i in range(n,(n*10)+1,n):
    print(n,'*',int(i/n),'=',i)
```

```python
# matrix
for i in range(1,6):
    for j in range(1,6):
        print((i,j),end=' ')
    print()
```

```python
c = 0
for i in range(1,6):
    for j in range(1,6):
        if i+j == 6:
            c += 1
print("Number of pairs that sum to 6:", c)
print("The percentage of pairs that sum to 6 is:", (c/(c*c))*100, "%")
```

```python
for n in range(1,13):
    c = 0
    for i in range(1,7):
        for j in range(1,7):
            if i+j == n:
                c += 1
    print(n,'=',round((c/36)*100,2),'%')
```
