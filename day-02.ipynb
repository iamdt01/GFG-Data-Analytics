# Strings and their operations

```python
x = "Geeks For Geeks"
```

```python
# concatenate string
print(x + " is a computer science portal for geeks")
```

```python
# replicate string
print(x * 2)
```

```python
# length of string
print(len(x))
```

```python
# String slicing
print(x[0:5]) # prints 'Geeks'
print(x[:5])
print(x[6:]) # prints 'For Geeks'
print(x[-1])
```

```python
# case conversion
print(x.upper()) # prints 'GEEKS FOR GEEKS'
print(x.lower()) # prints 'geeks for geeks'
print(x.title()) # prints 'Geeks For Geeks'
print(x.capitalize()) # prints 'Geeks for geeks'
print(x.swapcase()) # prints 'gEEKS fOR gEEKS'
print(x.casefold()) # prints 'geeks for geeks' #casefold() is similar to lower() but more aggressive in terms of converting characters to lowercase.
print(x.isupper()) # prints 'False'
print(x.islower()) # prints 'False'
print(x.istitle()) # prints 'True'
print(x.isalpha()) # prints 'False' #isalpha() returns True if all characters in the string are alphabetic and there is at least one character, otherwise False.
print(x.isdigit()) # prints 'False' #isdigit() returns True if all characters in the string are digits and there is at least one character, otherwise False.
print(x.isalnum()) # prints 'False' #isalnum() returns True if all characters in the string are alphanumeric and there is at least one character, otherwise False.
print(x.isnumeric()) # prints 'False' #isnumeric() returns True if all characters in the string are numeric and there is at least one character, otherwise False.
print(x.isdecimal()) # prints 'False' #isdecimal() returns True if all characters in the string are decimal and there is at least one character, otherwise False.
```

```python
print('a'.isalpha())
print('a '.isalpha())
print('a'.isalnum())
print('1'.isnumeric())
print('1'.isdigit())
```

```python
print(x[0])
print(x[-1])
```

```python
print(x[len(x)]) # error: string index out of range
```

```python
print(x[::-1]) #reverse a string
print(x[::2]) # prints 'Gek orek' #prints every second character in the string
print(x[::])
```

```python
# stripping
print(x.strip()) # prints 'Geeks For Geeks' #strip() removes leading and trailing whitespace characters from the string.
print('   a b c    '.strip())
```

```python
# trimming
print(x.lstrip()) # prints 'Geeks For Geeks' #lstrip() removes leading whitespace characters from the string.
print('   a b c    '.lstrip())
print(x.rstrip()) # prints 'Geeks For Geeks' #rstrip() removes trailing whitespace characters from the string.
print('   a b c    '.rstrip())
```

```python
# replacing
print(x.replace("Geeks", "GFG")) # prints 'GFG For GFG' #replace() returns a copy of the string with all occurrences of a substring replaced with another substring.
```

```python
# count
print(x.count("Geeks")) # prints '2' #count() returns the number of occurrences of a substring in the string.
print('aaaabaababababbvvvc'.count('a')) # prints '9'
print('aaaabaababababbvvvc'.count('c')) # prints '1'
print("AAaaacbAAAaa".lower().count('a')) # prints '10' #lower() converts all characters in the string to lowercase.
```

```python
# startswith() and endswith()
print(x.startswith("Geeks")) # prints 'True' #startswith() returns True if the string starts with the specified prefix, otherwise False.
print(x.endswith("Geeks")) # prints 'True' #endswith() returns True if the string ends with the specified suffix, otherwise False.
```

```python
# Centering
print(x.center(30, '*')) # prints '*******Geeks For Geeks*******' #center() returns a new string that is centered in a field of a given width. The original string is padded with the specified fill character (default is space) on both sides to achieve the desired width.
print(x.ljust(30, '*')) # prints 'Geeks For Geeks***************' #ljust() returns a new string that is left-justified in a field of a given width. The original string is padded with the specified fill character (default is space) on the right side to achieve the desired width.
print(x.rjust(30, '*')) # prints '***************Geeks For Geeks' #rjust() returns a new string that is right-justified in a field of a given width. The original string is padded with the specified fill character (default is space) on the left side to achieve the desired width.
```

# Variables

```python
# assigining a value to a variable
a = 10
b = 5
print(a + b) # prints '15'
print(a - b) # prints '5'
print(a * b) # prints '50'
print(a / b) # prints '2.0'
print(a // b) # prints '2' #floor division
print(a % b) # prints '0' #modulus
print(a ** b) # prints '100000' #exponentiation
```

```python
# assigining multiple values to multiple variables
a, b, c = 10, 5, 2
print(a,b,c)
```

```python
a = b = c = 0
print(a,b,c)
```

```python
e = [1,2,3,4]
a,b,c,d = e
print(e)
print(type(e))
print(a,b,c,d)
print(type(a), type(b), type(c), type(d))
```

```python
# Taking input from user
user_input_1 = int(input("Enter the first number: "))
user_input_2 = int(input("Enter the second number: "))
print("You entered:", user_input_1, 'and', user_input_2)
print('Sum : ', user_input_1 + user_input_2)
print('Difference : ', user_input_1 - user_input_2)
print('Product : ', user_input_1 * user_input_2)
print('Quotient : ', user_input_1 / user_input_2)
print('Floor Division : ', user_input_1 // user_input_2)
print('Modulus : ', user_input_1 % user_input_2)
print('Exponentiation : ', user_input_1 ** user_input_2)
```

# Variable Naming Convention

```python
# snake casing is a naming convention where words are separated by underscores (_) and all letters are in lowercase. It is commonly used in programming languages like Python for variable and function names. For example, instead of writing "myVariableName", you would write "my_variable_name".
my_variable_name = "This is an example of snake casing."
# pascal casing is a naming convention where the first letter of each word is capitalized and there are no spaces or underscores between words. It is commonly used in programming languages like C# and Java for class names. For example, instead of writing "myVariableName", you would write "MyVariableName".
MyVariableName = "This is an example of pascal casing."
# camel casing is a naming convention where the first letter of the first word is lowercase and the first letter of each subsequent word is capitalized, with no spaces or underscores between words. It is commonly used in programming languages like JavaScript for variable and function names. For example, instead of writing "my_variable_name", you would write "myVariableName".
myVariableName = "This is an example of camel casing."
```
