# Functions

```python
# function is a set of code to do a certain work repeatedly without writing the code multiple times
# def() is used to define a function
def hello(): # function definition
    print('Hello')

hello() # functiuon call
hello() # unparameterized function
```

```python
# parsing a parameter in a function
def hello(client_name):
    print('Hello,',client_name,'Good Morning!')

hello('DT') # parameterized function
```

```python
# variables declared inside the function don't have the scope of being used outside the function
g_val = 10
def test():
    l_val = 5
    print(g_val,l_val)
test()
print(g_val,l_val)
```

```python
g_val = 10
def test():
    l_val = 5
    print(g_val,l_val)
test()
```

```python
def greet(*args):
    for i in args:
        print('Hello',i)
```

```python
greet()
```

```python
greet('DT')
```

```python
def greet(n = '-----'):
    print('Hello',n)
```

```python
greet()
greet(n = 'DT')
```

```python
def s(a,b):
    return a+b
```

```python
x = s(5,10)
print(x)
print(s(1,3))
```

```python
def s(a,b):
    return a+b, a*b, a**b
```

```python
x = (s(2,3))
print(x)
for i in x:
    print(i)
d,e,f = x # unpacking
print(d)
print(e)
print(f)
```

```python
# square
def sq(lst):
    return [i**2 for i in lst]

def cu(lst):
    return [i**3 for i in lst]
```

```python
x = sq([1,2,3,4])
y = cu([1,2,3,4])
print(x)
print(y)
```

```python
def final():
    lst1 = sq([1,2,3,4])
    lst2 = cu([1,2,3,4])
    return [lst1[i] + lst2[i] for i in range(len(lst1))]
```

```python
print(final())
```
