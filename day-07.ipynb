# Data Structures 
# List


```python
l = [1,2,3,4,5,6.7,8.9]
print(l)
```

```python
l = ['a','b','c','d']
print(l)
```

```python
l = ['DT', 'is' , 'The', ' KING']
print(l)
```

```python
l = [1,2,3,4,'a','b','c','d',1.1,2.3,4.5,True,False,'DT','Is','The','KING']
print(l)
```

```python
for i in l:
    print(i,end=',')
```

```python
print(l[0],',',l[-1])
```

```python
print(l[::-1])
```

```python
print(l[1:6:2])
```

```python
print(l[-1][0])
```

```python
# list is mutuable i.e. we can change or update the list at any index
```

```python
l[-1] = 'LORD'
print(l)
```

```python
print(l[::-2])
```

```python
# appending to a list i.e adding to a list
l.append('Hanuman')
print(l)
```

```python
# removing from list
l.remove(1)
print(l)
```

```python
l.insert(0,1) # adding at a particular index
print(l)
```

```python
# length of a list
print(len(l))
```

```python
# sorting alist
l1 = [1,5,3,0,24,21]
print(sorted(l1))
```

```python
# finding an element
print(l.index('LORD'))
```

```python
# counting the elements
print(l.count('DT'))
```

```python
# extending a list is appending many elements at a single time
l.extend([1,2,3,5])
print(l)
```

```python
# max and min of a list
print(min(l1))
print(max(l1))
```

```python
# for loop
for i in l1:
    print(i, end=' ')
```

```python
# while loop
i = 0
while i<len(l1):
    print(l1[i],end=' ')
    i+=1
```

```python
# for loop
for i in range(len(l1)):
    print(i,'--->',l1[i])
```

```python
# accessing elments in a reversed manner
for i in range(len(l1)-1,-1,-1):
    print(i,'--->',l1[i])
```

# Multidimensional List

```python
ml = [[1,2],['a','b'],['DT'],'GOD','ABOVE','ALL']
```

```python
print(ml)
```

```python
# a list containing multiple one dimensional lists
```

```python
for i in ml:
    print(i)
```

```python
print(ml[::-1])
```

```python
print(ml[0])
```

```python
print(ml[0][0])
```

```python
ml[0][0] = 16
print(ml)
```

```python
# delete from a list
del ml[0]
print(ml)
```

```python
print(ml[::-1])
```

```python
print(ml)
```

```python
ml.reverse()
```

```python
print(ml[::-1])
```

```python
for i in ml:
    if type(i) == type(list()):
        for j in i:
            print(j)
    else:
        print(i)
```
