# Multi-dimensional Dictionary

```python
d = {1:{'name':'DT','ph':1234567890,'marks':{'hin':42,'eng':45,'sci':41}},
     2:{'name':'DT1','ph':1234567890,'marks':{'hin':44,'eng':43,'sci':49}}}
```

```python
print(d)
```

```python
for i in d.keys():
    print(i,d[i]['name'],d[i]['marks'])
```

```python
d[1]['name'] = 'King'
print(d)
```

```python
d[3] = {'name':'DT','ph':1234567890,'marks':{'hin':42,'eng':45,'sci':41}}
print(d)
```

```python
del d[3]
print(d)
```
