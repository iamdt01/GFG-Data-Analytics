# Dictionary Comprehension

```python
d = {k:k**2 for k in range(1,4)}
print(d)
```

```python
d = {i:i**2 for i in range(1,11) if i%2==0}
print(d)
```

```python
l = ['apples','bananas','mangoes']
d = {i:len(i) for i in l}
print(d)
```

```python
d = {'num_'+str(i) : i for i in range(1,11)}
print(d)
```

```python
d = {k:v for k,v in d.items() if v%3==0}
print(d)
```

```python
l1 = ['a','b','c']
l2 = 1,2,3
d = {l2[j]:l1[j] for j in range(len(l1))}
print(d)
```

```python
m = [[1,2,3],[4,5,6],[7,8,9]]
d = {(i,j):m[i][j] for i in range(3) for j in range(3)}
print(d)
```
