# Operators in Python

```python
'''Operators in Python
1. Arithmetic Operators: +, -, *, /, %, **, //
2. Comparison Operators: ==, !=, >, <, >=, <=
3. Assignment Operators: =, +=, -=, *=, /=, %=, **=, //=
4. Logical Operators: and, or, not
5. Bitwise Operators: &, |, ^, ~, <<, >>
6. Membership Operators: in, not in
7. Identity Operators: is, is not
8. Operator Precedence: The order in which the operators are evaluated in an expression. The order of precedence is as follows:
    1. Parentheses ()
    2. Exponentiation (**)
    3. Multiplication (*), Division (/), Floor Division (//), Modulus (%)
    4. Addition (+), Subtraction (-)
'''
```

```python
# addition
print(2+3)
print(2.5+3.5)
print(True + True)
print(True + False)
print(False + False)
print('Hello' + ' '+ 'World')
```

```python
# subtraction
print(2-3)
print(2.5-3.5)
print(True - True)
print(True - False)
print(False - True)
print(False - False)
```

```python
# Multiplication
print(2*3)
print(2.5*3.5)
print(True * True)
print(True * False)
print(False * False)
print('Hello '*3)
```

```python
# Divison
print(2/3)
print(2.5/3.5)
print(True / True)
```

```python
# Floor Divison (When the integer division is performed, the result is rounded down to the nearest integer)
print(2//3)
print(2.5//3.5)
print(True // True)
```

```python
# Modulus (The modulus operator returns the remainder of the division)
print(2%3)
print(3.5%2.5)
print(True // True)
```

```python
# Exponentiation (The exponentiation operator returns the result of raising the first operand to 
# the power of the second operand)
print(2**3)
print(3.5**2.5)
```

```python
# abs (the abs() function returns the absolute value of a number. 
# The absolute value of a number is the distance between the number and zero on the number line. 
# The absolute value of a number is always positive.)
print(abs(2))
print(abs(-2))
```

```python
# round (the round() function returns the nearest integer to a given number.
# If the fractional part of the number is 0.5 or greater, the function rounds the number up to the next integer. If the fractional part is less than 0.5, the function rounds the number down to the previous integer.)
print(round(2.567, 2))
print(round(2.567, 1))
```

```python
# power (the pow() function returns the value of a number raised to the power of another number.)
print(pow(2, 3))
```

```python
# Comparison
print(2==3)
print(2!=3)
print(2>3)
print(2<3)
print(1.0 == 1)
print(True is True)
print(True == 1)
print(False == 0)
print(2<=3)
print(2>=3)
```

```python
# logical Operators
print(True and True)
print(True and False)
print(False and True)
print(False and False)
print(True or True)
print(True or False)
print(False or True)
print(False or False)
print(not True)
print(not False)
print(True and True and False)
print(True or True or False)
print(not True and False)
print(not True or False)
print(not True and not False)
print(not True or not False)
```
